```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("profit_sales_analysis_dataset.csv")
df
```

```python
numeric_cols=["Unit_Price","Quantity","Discount_Percent","Sales","Profit","Profit_Margin_Percent","Cost"]
deepika=df[numeric_cols].corr()
deepika
```

```python
#heat map
sns.heatmap(deepika,
            annot=True,
            cmap="coolwarm",
            cbar=False,
            square=True,
            linewidth=1)
plt.title("profit_margin")
plt.show()
```

```python
#bar chart
sns.barplot(x="Category",y="Sales",data=df,width=0.5,color="blue")
plt.title("Sales v/s Category")
plt.show()
```

```python
#line chart
sns.lineplot(x="Category",y="Profit",data=df)
plt.title("Profit v/s Category")
plt.show()
```

```python
#budget vs actual charts,
data={
    "department":["HR","finace","Sales","Marketing","IT","Production","Kannada","Hindi"],
    "Budget":[70000,80000,60000,40000,30000,20000,10000,90000],
    "Actual":[54000,60000,85000,35000,75000,55000,98000,43000]
}

df=pd.DataFrame(data)
print(df)

df_long=df.melt(
    id_vars="department",
    value_vars=["Budget","Actual"],
    var_name="Type",
    value_name="Amount"
)

df_long

sns.barplot(
    data=df_long,
    x="department",
    y="Amount",
    hue="Type"
)
plt.title("Budget vs Actual")
plt.xlabel("department")
plt.ylabel("Amount")
plt.show()
```
