```python
try:
    balance=int(input("Enter Cash Book Balance:"))
    n=int(input("Enter the number of Transactions:"))

    
    Transactions=[]
    for i in range(n):
        print("\n transaction",i+1)
        try:
            amount=int(input("Enter the Amount:"))
            t_type=input("debit or credit:")
            Transactions.append([amount,t_type])
        except ValueError:
            print("invaild input, enter vaild input")
    for transaction in Transactions:
        amount=transaction[0]
        t_type=transaction[1]
        
        if t_type.lower()=="credit":
            balance=balance+amount
        elif t_type.lower()=="debit":
            balance=balance-amount
        else:
            print("invaid input")

    print("FINAL RECONCILED BALANCE:",balance)
except ValueError:
    print("invaild input,enter vaild input")
```
